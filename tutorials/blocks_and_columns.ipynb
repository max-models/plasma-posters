# Blocks and columns

A poster is a title, the authors and blocks of text and figures in columns. Text, titles and captions are LaTeX, so they can hold maths.

````python
from plasma_posters import Poster

poster = Poster(
    "Speeds in a thermal plasma",
    authors="Max Lindqvist",
    institute="Max Planck Institute for Plasma Physics",
    columns=2,
)
poster.block("The distribution", r"$f(v) \propto v^2 \exp(-m v^2 / 2 k_B T)$")
poster.figure(
    "Three temperatures", "maxwellian.pdf", caption="Normalised to its peak.", column=1
)
````

Blocks are stacked in their column in the order they are added. `to_latex` gives the document, a [tikzposter](https://ctan.org/pkg/tikzposter) file:

````python
print(poster.to_latex())
````

`poster.save("poster.tex")` writes it, and `poster.build("poster.pdf")` writes the `.tex` next to the PDF and compiles it with lualatex. Figure paths are relative to the `.tex` file. A `.tex` edited by hand is compiled with

```bash
plasma-posters build poster.tex
```
