# Examples

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Blocks and columns](/plasma-posters/examples/blocks-and-columns/) builds a poster and shows its LaTeX.
- [Scripts](/plasma-posters/example-scripts/) are complete Python scripts with their plots.
- [Building the docs](/plasma-posters/development/docs/) explains how to add an example and control its output.
